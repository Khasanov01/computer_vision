```python
import cv2
```

```python
from google.colab.patches import cv2_imshow

image= cv2.imread("/content/images.jpg")
cv2_imshow(image)

print('shape', image.shape)
#height=> 553
#width=>555
#colour channel=>3(RGB)  Red blue Green  => BGR in open_cv

print('height', image.shape[0])
print('width', image.shape[1])
print('channels', image.shape[2])

```

```python
(b, g, r)=image[100, 50]
print("blue", b,  "green", g,  "red", r)
```

```python
gray=cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized=cv2.resize(image, (300, 300))
cv2_imshow(resized)
```

```python
cropped=gray[100:400, 150:450]
cv2_imshow(cropped)
```

```python
rotated=cv2.rotate(resized, cv2.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
flipped=cv2.flip(gray, 1)
cv2_imshow(flipped)
```

```python
cv2.imwrite=("flipped_cat.jpg", flipped)
```
